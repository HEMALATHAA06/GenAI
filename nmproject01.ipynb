```python
from transformers import pipeline
import random
import matplotlib.pyplot as plt

# AI pipeline
generator = pipeline("text-generation", model="gpt2")

# Mood input
mood = input("Enter your current mood: ").strip().lower()

# Prompt
prompt = f"Write a beautiful poem that expresses the feeling of being {mood}."

# Generate
poem = generator(prompt, max_length=100, num_return_sequences=1)[0]['generated_text']

# Color map (optional visual)
mood_colors = {
    'happy': '#ffe066',
    'sad': '#6fa8dc',
    'anxious': '#ff6961',
    'calm': '#a3d9a5',
    'nostalgic': '#d4a5a5'
}
bg_color = mood_colors.get(mood, '#ffffff')

# Display
plt.figure(figsize=(10,2))
plt.text(0.5, 0.5, poem, wrap=True, ha='center', va='center', fontsize=12)
plt.axis('off')
plt.gca().set_facecolor(bg_color)
plt.show()
```
