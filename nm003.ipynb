```python
import gradio as gr
from transformers import pipeline

# Load a pre-trained sentiment analysis model
sentiment_analyzer = pipeline("sentiment-analysis")

def analyze_sentiment(text):
    # Analyze the sentiment of the input text
    result = sentiment_analyzer(text)[0]
    label = result['label']
    score = round(result['score'], 4)
    return f"Sentiment: {label}", f"Confidence: {score}"

# Create a Gradio interface
interface = gr.Interface(
    fn=analyze_sentiment,
    inputs=gr.Textbox(
        lines=2,
        placeholder="Enter text to analyze sentiment...",
        label="Input Text"
    ),
    outputs=[
        gr.Textbox(label="Sentiment Label"),
        gr.Textbox(label="Confidence Score")
    ],
    title="Sentiment Analysis App",
    description="Enter text to determine its sentiment (Positive/Negative) and confidence score.",
    examples=[
        ["I love this product!"],
        ["This is the worst experience I've ever had."],
        ["The movie was okay, but not great."]
    ]
)

# Launch the Gradio interface
interface.launch()
```

```python
import gradio as gr
import spacy

# Load the English language model
nlp = spacy.load("en_core_web_sm")

def pos_tagger(text):
    """
    Performs part-of-speech tagging on the input text using spaCy.

    Args:
        text (str): The input text to be tagged.

    Returns:
        list: A list of tuples, where each tuple contains a word and its POS tag.
    """
    doc = nlp(text)
    results = [(token.text, token.pos_) for token in doc]
    return results

# Create the Gradio interface
iface = gr.Interface(
    fn=pos_tagger,
    inputs=gr.Textbox(lines=3, placeholder="Enter text here..."),
    outputs=gr.HighlightedText(
        label="POS Tags",
        color_map={
            "NOUN": "blue",
            "VERB": "red",
            "ADJ": "green",
            "ADV": "yellow",
            "PROPN": "purple",
            "ADP": "orange",
            "DET": "cyan",
            "PRON": "pink",
            "CCONJ": "gray",
            "SCONJ": "brown",
            "INTJ": "teal",
            "NUM": "violet",
            "SYM": "lime",
            "PUNCT": "white",
            "SPACE": "black",
            "X": "silver",
        },
    ),
    title="Part-of-Speech Tagger",
    description="Enter a sentence or paragraph to see its part-of-speech tags."
)

# Launch the interface
iface.launch()
```

```python
import gradio as gr
from transformers import pipeline

# Load the translation model
translator = pipeline("translation", model="Helsinki-NLP/opus-mt-en-es")

def translate_text(text):
    """
    Translates English text to Spanish.

    Args:
        text (str): The English text to be translated.

    Returns:
        str: The translated Spanish text.
    """
    translated_text = translator(text)[0]['translation_text']
    return translated_text

# Create the Gradio interface
iface = gr.Interface(
    fn=translate_text,
    inputs=gr.Textbox(lines=3, placeholder="Enter English text here..."),
    outputs=gr.Textbox(label="Spanish Translation"),
    title="English to Spanish Translator",
    description="Translate English text to Spanish using the Helsinki-NLP model."
)

# Launch the interface
iface.launch()
```

```python
import gradio as gr
from transformers import pipeline

# Load the summarization pipeline
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

def summarize_text(text):
    """
    Summarizes the input text using the BART model.

    Args:
        text (str): The input text to be summarized.

    Returns:
        str: The summarized text.
    """
    summary = summarizer(text, max_length=130, min_length=30, do_sample=False)[0]['summary_text']
    return summary

# Create the Gradio interface
iface = gr.Interface(
    fn=summarize_text,
    inputs=gr.Textbox(lines=5, placeholder="Enter text to summarize..."),
    outputs=gr.Textbox(label="Summary"),
    title="Text Summarization App",
    description="Enter a long text to generate a concise summary using the BART model."
)

# Launch the interface
iface.launch()
```

```python
import gradio as gr
from transformers import pipeline

# Load the question answering model
question_answerer = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")

def answer_question(context, question):
    """
    Answers a question based on a given context.

    Args:
        context (str): The context text to extract the answer from.
        question (str): The question to be answered.

    Returns:
        str: The answer extracted from the context.
    """
    result = question_answerer(question=question, context=context)
    return result['answer']

# Create the Gradio interface
iface = gr.Interface(
    fn=answer_question,
    inputs=[
        gr.Textbox(lines=7, placeholder="Enter context here...", label="Context"),
        gr.Textbox(placeholder="Enter your question here...", label="Question")
    ],
    outputs=gr.Textbox(label="Answer"),
    title="Question Answering App",
    description="Enter a context and a question to get an answer."
)

# Launch the interface
iface.launch()
```

```python
import gradio as gr
from bs4 import BeautifulSoup

def extract_website_info(html_file_content, query):
    """
    Extracts text from HTML and returns relevant information based on the query.

    Args:
        html_file_content (str): The content of the HTML file.
        query (str): The search query.

    Returns:
        str: Relevant information extracted from the HTML or a message
             if no relevant information is found.
    """
    soup = BeautifulSoup(html_file_content, 'html.parser')

    # Extract all text from the HTML
    all_text = soup.get_text(separator=' ', strip=True)

    # Perform basic information retrieval (keyword matching)
    if query.lower() in all_text.lower():
        return f"Found relevant information:\n\n{all_text}"
    else:
        return "No relevant information found for the given query."


def process_and_extract(query, file_path):
    """
    Read the file and do information extraction
    """

    try:
        with open(file_path, 'r') as f:
            html_content = f.read()
            return extract_website_info(html_content, query)
    except FileNotFoundError:
        return "HTML file not found."
    except Exception as e:
        return f"An error occurred: {e}"

# Create the Gradio interface
iface = gr.Interface(
    fn=process_and_extract,
    inputs=[
        gr.Textbox(lines=1, placeholder="Enter your query...", label="Query"),
        gr.Textbox(lines=1, placeholder="Path to the HTML file...", label="HTML File path")
    ],
    outputs=gr.Textbox(label="Retrieved Information"),
    title="Information Retrieval from Photo Craft Website",
    description="Enter a query to search for specific information within the Photo Craft website's HTML content."
)

# Launch the interface
iface.launch()
```

```python
import gradio as gr
from transformers import pipeline

# Load the text generation pipeline
generator = pipeline('text-generation', model='gpt2')

def generate_tagline(topic):
    """
    Generates a tagline for the given topic using the GPT-2 model.

    Args:
        topic (str): The main topic or keywords related to the tagline.

    Returns:
        str: The generated tagline.
    """
    prompt = f"A catchy tagline for Photo Craft, an online photo editor: {topic}"
    generated_text = generator(prompt, max_length=50, num_return_sequences=1)[0]['generated_text']
    return generated_text

# Create the Gradio interface
iface = gr.Interface(
    fn=generate_tagline,
    inputs=gr.Textbox(lines=1, placeholder="Enter a topic or keyword...", label="Topic"),
    outputs=gr.Textbox(label="Generated Tagline"),
    title="Tagline Generator for Photo Craft",
    description="Enter a topic related to online photo editing and generate a catchy tagline."
)

# Launch the interface
iface.launch()
```
