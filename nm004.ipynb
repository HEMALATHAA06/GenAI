```python
import os
import google.generativeai as genai
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

# Get API key from .env file
API_KEY = os.getenv("GEMINI_API_KEY")

# Initialize Gemini AI
genai.configure(api_key=API_KEY)

def chat_with_gemini():
    print("Chatbot: Hello! Type 'exit' to end the conversation.")

    while True:
        user_input = input("You: ")

        if user_input.lower() == "exit":
            print("Chatbot: Goodbye!")
            break

        try:
            # Send request to Gemini API
            model = genai.GenerativeModel("gemini-pro")  # Use "gemini-pro" or another available model
            response = model.generate_content(user_input)

            # Print chatbot response
            print("Chatbot:", response.text)

        except Exception as e:
            print(f"Error: {e}")

if __name__ == "__main__":
    chat_with_gemini()
```

```python
import requests
from bs4 import BeautifulSoup
from transformers import pipeline

# Load Hugging Face summarization model (open-source, no API key needed)
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

def fetch_alabama_news():
    """Fetch latest Alabama news from a free RSS feed."""
    RSS_FEED_URL = "https://rss.nytimes.com/services/xml/rss/nyt/US.xml"  # Example RSS feed

    response = requests.get(RSS_FEED_URL)
    if response.status_code != 200:
        print(f"❌ Failed to fetch news: {response.status_code}")
        return []

    soup = BeautifulSoup(response.content, "xml")
    items = soup.find_all("item")[:5]  # Get top 5 articles

    articles = []
    for item in items:
        title = item.title.text
        description = item.description.text
        link = item.link.text
        articles.append({"title": title, "content": description, "url": link})

    return articles

def summarize_text(text):
    """Summarize news using an open-source AI model."""
    if not text or len(text.split()) < 50:  # Skip short texts
        return text

    summary = summarizer(text, max_length=50, min_length=20, do_sample=False)
    return summary[0]['summary_text']

def main():
    """Fetch and summarize news without API keys."""
    print("\n📰 **Latest Alabama & US News Summaries (Free Source)**\n")

    news_articles = fetch_alabama_news()
    if not news_articles:
        print("No news found.")
        return

    for i, article in enumerate(news_articles, start=1):
        title = article["title"]
        content = article["content"]
        url = article["url"]

        # Generate summary
        summary = summarize_text(content)

        print(f"\n🔹 **{i}. {title}**")
        print(f"📜 Summary: {summary}")
        print(f"🔗 Read more: {url}")

if __name__ == "__main__":
    main()
```

```python
from transformers import MarianMTModel, MarianTokenizer

def load_model(src_lang, tgt_lang):
    """Load the translation model for the given language pair."""
    model_name = f"Helsinki-NLP/opus-mt-{src_lang}-{tgt_lang}"
    tokenizer = MarianTokenizer.from_pretrained(model_name)
    model = MarianMTModel.from_pretrained(model_name)
    return model, tokenizer

def translate_text(text, src_lang="en", tgt_lang="fr"):
    """Translate text from source language to target language."""
    model, tokenizer = load_model(src_lang, tgt_lang)

    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True)
    translated = model.generate(**inputs)
    result = tokenizer.decode(translated[0], skip_special_tokens=True)

    return result

def main():
    print("\n🌍 AI-Powered Language Translator")
    src_lang = input("Enter source language code (e.g., 'en' for English): ").strip()
    tgt_lang = input("Enter target language code (e.g., 'fr' for French): ").strip()
    text = input("Enter text to translate: ").strip()

    if not text:
        print("⚠️ No text provided!")
        return

    translation = translate_text(text, src_lang, tgt_lang)
    print(f"\n🔹 Translated Text ({src_lang} → {tgt_lang}): {translation}")

if __name__ == "__main__":
    main()
```
